```python
import matplotlib.pyplot as plt
import numpy as np

# Dados convertidos para minutos (exemplo para SAMPLE1)
labels = ['real', 'user', 'sys']
bwa_mem = [139.59, 129.48, 19.31]  # SAMPLE1
bwa_mem2 = [87.76, 83.98, 12.15]    # SAMPLE1
bowtie = [194.98, 392.02, 12.33]    # SAMPLE1

angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]  # Fechar o círculo

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'polar': True})

# Adicionar dados
def plot_radar(data, color, label):
    data += data[:1]
    ax.plot(angles, data, color=color, linewidth=2, label=label)
    ax.fill(angles, data, color=color, alpha=0.1)

plot_radar(bwa_mem, 'blue', 'bwa-mem')
plot_radar(bwa_mem2, 'red', 'bwa-mem2')
plot_radar(bowtie, 'green', 'bowtie')

# Configurações
ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels)
ax.set_title('Comparison of Times (SAMPLE1)', pad=20)
ax.legend(loc='upper right')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados da tabela
softwares = ['GATK', 'Freebayes', 'bcftools', 'DeepVariant']
amostras = ['SAMPLE1', 'SAMPLE2', 'SAMPLE3', 'average']

# Tempos em minutos (convertidos de formato m+s para minutos decimais)
tempos = {
    'SAMPLE1': [52 + 52.502/60, 25 + 23.899/60, 20 + 9.576/60, 6 + 49.238/60],
    'SAMPLE2': [39 + 32.706/60, 20 + 27.350/60, 14 + 59.843/60, 4 + 59.694/60],
    'SAMPLE3': [55 + 30.946/60, 25 + 56.136/60, 17 + 28.340/60, 5 + 33.317/60],
    'average': [49.3119, 25.5965, 17.5431, 5.7903]
}

# Configuração do gráfico
x = np.arange(len(softwares))  # localização dos labels
width = 0.2  # largura das barras
multiplier = 0

fig, ax = plt.subplots(figsize=(12, 6))

# Criando as barras para cada amostra
for amostra in amostras:
    offset = width * multiplier
    rects = ax.bar(x + offset, tempos[amostra], width, label=amostra)
    multiplier += 1

# Adicionando detalhes ao gráfico
ax.set_ylabel('Execution time (minutes)')
ax.set_title('Runtime comparison between software')
ax.set_xticks(x + width*1.5)
ax.set_xticklabels(softwares)
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
ax.grid(True, axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
